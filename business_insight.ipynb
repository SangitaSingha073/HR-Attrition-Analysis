# Draw business insights

```python
import pandas as pd
```

```python
#Load the dataset
df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
df['Attrition_Num'] = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
total = len(df)
left = df[df['Attrition'] == 'Yes'].shape[0]
rate = (left / total) * 100
print(f"Total Employees   : {total}")
print(f"Employees Left    : {left}")
print(f"Attrition Rate    : {rate:.1f}%")
```

```python
ot = df.groupby('OverTime')['Attrition_Num'].mean() * 100
print(ot)
```

```python
dept = df.groupby('Department')['Attrition_Num'].mean() * 100
print(dept.sort_values(ascending=False))
```

```python
sat = df.groupby('JobSatisfaction')['Attrition_Num'].mean() * 100
print(sat)
```

```python
income = df.groupby('Attrition')['MonthlyIncome'].mean()
print(income)
```

```python
df['AgeGroup'] = pd.cut(df['Age'],
                  bins=[18, 25, 35, 45, 60],
                  labels=['18-25', '26-35', '36-45', '46-60'])

age_attr = df.groupby('AgeGroup')['Attrition_Num'].mean() * 100
print(age_attr)
```

```python
mar = df.groupby('MaritalStatus')['Attrition_Num'].mean() * 100
print(mar.sort_values(ascending=False))
```

```python
df['TenureGroup'] = pd.cut(df['YearsAtCompany'],
                    bins=[0, 2, 5, 10, 40],
                    labels=['0-2 yrs', '3-5 yrs', '6-10 yrs', '10+ yrs'])

tenure = df.groupby('TenureGroup')['Attrition_Num'].mean() * 100
print(tenure)
```
